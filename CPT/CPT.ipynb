```python
from pathlib import Path
import pandas as pd

folder_data = Path("data")
csv = []

for file in folder_data.glob("*.csv"):
    file_csv = pd.read_csv(file)
    csv.append(file_csv)

df = pd.concat(csv)
```

```python
df
```

```python
valid = df[df["output_nonformal"] != "ERROR_API"]
```

```python
valid
```

```python
valid.to_csv("sudah_nonformal.csv", index=False)
```

```python
referensi = pd.read_csv("referensi.csv")
referensi
```

```python
berita_set = set(valid["text_asli"])
belum_keliatan = []

for data in referensi["text"]:
    # print(data)
    if data not in berita_set:
        belum_keliatan.append(data)

df_hasil = pd.DataFrame(belum_keliatan, columns=["text"])

df_hasil.to_csv("referensi2.csv", index=False)
```

```python
# bisa gini juga ya biar lebih cepet misahin pake pandas
kondisi = ~referensi["text"].isin(berita_set)

belum = referensi.loc[kondisi, "text"].tolist()
```

```python
# cek aja takut salah
panjang_valid = len(valid)
panjang_referensi = len(referensi)

print(f"panjang data valid = {panjang_valid}")
print(f"panjang data referensi = {panjang_referensi}")
print(f"panjang data baru harusnya = {panjang_referensi} - {panjang_valid}: {panjang_referensi - panjang_valid}")

output = pd.read_csv("referensi2.csv")
print(f"panjang data baru beneran = {len(output)}")
```
